# STAGE 2 DATA CLEANING & PREPROCESSING

```python
# CREATE CLEANING COPY

df_clean = df.copy()
print("Original Dataset Shape:", df.shape)
print("Cleaning Dataset Shape:", df_clean.shape)
```

```python
# STANDARDIZE COLUMN NAMES

df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("\nColumn Names:")
print(df_clean.columns.tolist())
```

```python
# HANDLE WHITESPACE IN STRING VALUES

object_columns = df_clean.select_dtypes(
    include="object"
).columns

for col in object_columns:

    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

```python
# HANDLE STRING NULL VALUES

null_values = [
    "null",
    "NULL",
    "Null",
    "nan",
    "NaN",
    "NAN",
    "none",
    "None",
    ""
]

df_clean.replace(
    null_values,
    np.nan,
    inplace=True
)
```

```python
print("\n========== MISSING VALUES BEFORE CLEANING ==========")

missing_before = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    )
}).sort_values(
    by="Missing_Count",
    ascending=False
)

display(missing_before)
```

```python
# HANDLE DUPLICATES

duplicate_count = df_clean.duplicated().sum()

print("\nDuplicate Rows:", duplicate_count)

df_clean = df_clean.drop_duplicates()

print("Duplicates Removed:", duplicate_count)
```

```python
# HANDLE BOOKING ID

if "Booking_ID" in df_clean.columns:

    df_clean["Booking_ID"] = (
        df_clean["Booking_ID"]
        .astype("string")
        .str.strip()
    )

    missing_booking_id = df_clean["Booking_ID"].isna().sum()

    print("\nMissing Booking IDs:", missing_booking_id)

    df_clean = df_clean.dropna(
        subset=["Booking_ID"]
    )
```

```python
# CONVERT DATE

if "Date" in df_clean.columns:

    df_clean["Date"] = pd.to_datetime(
        df_clean["Date"],
        errors="coerce",
        dayfirst=True
    )

    print("\nDate Column Converted!")
```

```python
# CONVERT TIME

if "Time" in df_clean.columns:

    df_clean["Time"] = pd.to_datetime(
        df_clean["Time"].astype("string"),
        errors="coerce"
    ).dt.strftime("%H:%M:%S")

    df_clean["Time"] = df_clean["Time"].replace(
        "NaT",
        np.nan
    )

    print("\nTime Column Standardized!")
```

```python
# CONVERT NUMERIC COLUMNS

numeric_columns = [
    "Booking_Value",
    "Ride_Distance",
    "Driver_Ratings",
    "Customer_Rating",
    "V_TAT",
    "C_TAT"
]

for col in numeric_columns:

    if col in df_clean.columns:

        df_clean[col] = pd.to_numeric(
            df_clean[col],
            errors="coerce"
        )

print("\nNumeric Columns Converted!")
```

```python
# CHECK MISSING VALUES

print("\n========== MISSING VALUES AFTER CONVERSION ==========")

display(
    df_clean.isnull().sum().to_frame(
        name="Missing_Count"
    )
)
```

```python
# VALIDATE NUMERIC VALUES

invalid_checks = {}

if "Booking_Value" in df_clean.columns:

    invalid_checks["Negative Booking Value"] = (
        df_clean["Booking_Value"] < 0
    ).sum()


if "Ride_Distance" in df_clean.columns:

    invalid_checks["Negative Ride Distance"] = (
        df_clean["Ride_Distance"] < 0
    ).sum()


if "Customer_Rating" in df_clean.columns:

    invalid_checks["Customer Rating Outside 0-5"] = (
        (
            (df_clean["Customer_Rating"] < 0)
            |
            (df_clean["Customer_Rating"] > 5)
        )
    ).sum()


if "Driver_Ratings" in df_clean.columns:

    invalid_checks["Driver Rating Outside 0-5"] = (
        (
            (df_clean["Driver_Ratings"] < 0)
            |
            (df_clean["Driver_Ratings"] > 5)
        )
    ).sum()


print("\n========== INVALID DATA CHECK ==========")

display(
    pd.DataFrame(
        invalid_checks.items(),
        columns=["Validation", "Count"]
    )
)
```

```python
# BOOKING STATUS CHECK

print("\n========== BOOKING STATUS ==========")

if "Booking_Status" in df_clean.columns:

    display(
        df_clean["Booking_Status"]
        .value_counts(dropna=False)
    )
```

```python
# VEHICLE TYPE CHECK

print("\n========== VEHICLE TYPE ==========")

if "Vehicle_Type" in df_clean.columns:

    display(
        df_clean["Vehicle_Type"]
        .value_counts(dropna=False)
    )
```

```python
# STANDARDIZE CATEGORICAL COLUMNS

categorical_columns = [
    "Booking_Status",
    "Vehicle_Type",
    "Payment_Method",
    "Pickup_Location",
    "Drop_Location"
]

for col in categorical_columns:

    if col in df_clean.columns:

        df_clean[col] = (
            df_clean[col]
            .astype("string")
            .str.strip()
        )
```

```python
print("\n========== CLEANED DATASET SHAPE ==========")

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
```

```python
print("\n========== FINAL MISSING VALUES ==========")

final_missing = pd.DataFrame({
    "Missing_Count": df_clean.isnull().sum(),
    "Missing_Percentage": (
        df_clean.isnull().sum() / len(df_clean) * 100
    )
}).sort_values(
    by="Missing_Count",
    ascending=False
)

display(final_missing)
```

```python
print("\n========== FINAL DUPLICATE CHECK ==========")

print(
    "Remaining Duplicates:",
    df_clean.duplicated().sum()
)
```

```python
print("\n========== FINAL DATA TYPES ==========")

display(
    df_clean.dtypes.to_frame(
        name="Data_Type"
    )
)
```

```python
# 20. SAVE CLEANED DATA

processed_path = os.path.join(
    PROJECT_DIR,
    "data",
    "processed"
)

os.makedirs(
    processed_path,
    exist_ok=True
)


# Save CSV

cleaned_csv_path = os.path.join(
    processed_path,
    "ola_cleaned.csv"
)

df_clean.to_csv(
    cleaned_csv_path,
    index=False
)


# Save Excel

cleaned_excel_path = os.path.join(
    processed_path,
    "ola_cleaned.xlsx"
)

df_clean.to_excel(
    cleaned_excel_path,
    index=False
)


print("\n========== FILES SAVED ==========")

print("CSV:", cleaned_csv_path)
print("Excel:", cleaned_excel_path)
```

# Stage 3 — EDA (Exploratory Data Analysis)

```python
# EDA DATA COPY

eda = df_clean.copy()

print("EDA Dataset Shape:", eda.shape)
```

```python
# FEATURE ENGINEERING

# Date

eda["Date"] = pd.to_datetime(
    eda["Date"],
    errors="coerce"
)


# Time

eda["Time"] = pd.to_datetime(
    eda["Time"].astype("string"),
    format="%H:%M:%S",
    errors="coerce"
)


# Hour

eda["Hour"] = eda["Time"].dt.hour


# Day

eda["Day"] = eda["Date"].dt.day


# Day Name

eda["Day_Name"] = eda["Date"].dt.day_name()


# Day of Week

eda["Day_of_Week"] = eda["Date"].dt.dayofweek


# Month

eda["Month"] = eda["Date"].dt.month_name()


# Weekend

eda["Weekend"] = np.where(
    eda["Day_of_Week"] >= 5,
    "Weekend",
    "Weekday"
)
```

```python
# BUSINESS KPI SUMMARY

total_bookings = eda["Booking_ID"].nunique()

successful_rides = (
    eda["Booking_Status"]
    .eq("Success")
    .sum()
)

cancelled_by_customer = (
    eda["Booking_Status"]
    .eq("Canceled by Customer")
    .sum()
)

cancelled_by_driver = (
    eda["Booking_Status"]
    .eq("Canceled by Driver")
    .sum()
)

driver_not_found = (
    eda["Booking_Status"]
    .eq("Driver Not Found")
    .sum()
)

total_revenue = eda.loc[
    eda["Booking_Status"] == "Success",
    "Booking_Value"
].sum()

avg_booking_value = eda.loc[
    eda["Booking_Status"] == "Success",
    "Booking_Value"
].mean()

avg_distance = eda.loc[
    eda["Booking_Status"] == "Success",
    "Ride_Distance"
].mean()

avg_customer_rating = eda.loc[
    eda["Booking_Status"] == "Success",
    "Customer_Rating"
].mean()

avg_driver_rating = eda.loc[
    eda["Booking_Status"] == "Success",
    "Driver_Ratings"
].mean()

print("\n========== BUSINESS KPI SUMMARY ==========")

print("Total Bookings:", total_bookings)
print("Successful Rides:", successful_rides)
print("Customer Cancellations:", cancelled_by_customer)
print("Driver Cancellations:", cancelled_by_driver)
print("Driver Not Found:", driver_not_found)
print("Total Revenue:", round(total_revenue, 2))
print("Average Booking Value:", round(avg_booking_value, 2))
print("Average Ride Distance:", round(avg_distance, 2))
print("Average Customer Rating:", round(avg_customer_rating, 2))
print("Average Driver Rating:", round(avg_driver_rating, 2))
```

```python
# BOOKING STATUS ANALYSIS

status_counts = (
    eda["Booking_Status"]
    .value_counts()
)

status_percentage = (
    eda["Booking_Status"]
    .value_counts(normalize=True) * 100
)

status_summary = pd.DataFrame({
    "Bookings": status_counts,
    "Percentage": status_percentage.round(2)
})

print("\n========== BOOKING STATUS ==========")

display(status_summary)


plt.figure(figsize=(10, 5))

sns.barplot(
    x=status_counts.index,
    y=status_counts.values
)

plt.title("Booking Status Distribution")
plt.xlabel("Booking Status")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

```python
# RIDE VOLUME OVER TIME

daily_bookings = (
    eda.groupby("Date")["Booking_ID"]
    .nunique()
    .sort_index()
)

plt.figure(figsize=(14, 5))

plt.plot(
    daily_bookings.index,
    daily_bookings.values,
    marker="o"
)

plt.title("Daily Ride Volume")
plt.xlabel("Date")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# BOOKINGS BY HOUR

hourly_bookings = (
    eda.groupby("Hour")["Booking_ID"]
    .nunique()
)

plt.figure(figsize=(12, 5))

sns.barplot(
    x=hourly_bookings.index,
    y=hourly_bookings.values
)

plt.title("Bookings by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Bookings")

plt.tight_layout()
plt.show()
```

```python
# BOOKINGS BY DAY OF WEEK

day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

weekday_bookings = (
    eda.groupby("Day_Name")["Booking_ID"]
    .nunique()
    .reindex(day_order)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=weekday_bookings.index,
    y=weekday_bookings.values
)

plt.title("Bookings by Day of Week")
plt.xlabel("Day")
plt.ylabel("Bookings")

plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
# VEHICLE TYPE ANALYSIS

vehicle_bookings = (
    eda.groupby("Vehicle_Type")["Booking_ID"]
    .nunique()
    .sort_values(ascending=False)
)

print("\n========== VEHICLE BOOKINGS ==========")

display(vehicle_bookings)


plt.figure(figsize=(10, 5))

sns.barplot(
    x=vehicle_bookings.index,
    y=vehicle_bookings.values
)

plt.title("Bookings by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Bookings")

plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```
